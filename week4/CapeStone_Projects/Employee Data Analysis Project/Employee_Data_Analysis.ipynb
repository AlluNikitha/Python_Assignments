# Capstone Project (Week 4): Employee Data Analysis

**Dataset:** Employee Dataset (Kaggle) — `employee_salary_dataset.csv`

**Tasks:**
- Load CSV using Pandas
- Calculate average salary, department count
- Filter employees above a salary threshold
- Export results to a new CSV

**Skill Gain:** Pandas, CSV handling, filtering.

## Load Data

```python
import pandas as pd

data = pd.read_csv("employee_salary_dataset.csv")
data
```

## Data Exploration

```python
data.info()
```

```python
data.head()
```

```python
data.shape
```

## Average Salary

```python
average_salary = data["Monthly_Salary"].mean()
print(f"Average Monthly Salary (all employees): {average_salary:.2f}")
```

## Department Count

```python
department_counts = data["Department"].value_counts()
department_counts
```

## Filter Employees Above Salary Threshold

```python
SALARY_THRESHOLD = 80000

high_earners = data[data["Monthly_Salary"] > SALARY_THRESHOLD]
print(f"Employees earning above {SALARY_THRESHOLD}: {len(high_earners)}")
high_earners
```

## Export Results to CSV

```python
high_earners.to_csv("high_earners.csv", index=False)
print("Results exported to 'high_earners.csv'.")
```
